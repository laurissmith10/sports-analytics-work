## Task 1 — Define the `Team` class

Write a class `Team` that:

- Takes a `name` in the constructor.
- Has attributes for: `played`, `wins`, `draws`, `losses`, `goals_for`, `goals_against` — all starting at 0.
- Has properties `points` (3 W + 1 D) and `goal_difference` (GF − GA).
- Has a method `add_match(scored, conceded)` that updates all the relevant attributes.

Use the class below as a starting point.

````python
class Team:
    def __init__(self, name):
        self.name = name
        # TODO Task 1 — initialise all counters
        self.played = 0
        self.wins = 0
        self.draws = 0
        self.losses = 0
        self.goals_for = 0
        self.goals_against = 0
        

    @property
    def points(self):
        # TODO Task 1 — compute points
        return self.wins * 3 + self.draws * 1
    
    @property
    def goal_difference(self):
        # TODO Task 1 — compute goal difference
        return self.goals_for - self.goals_against
    
    def add_match(self, scored, conceded):
        # TODO Task 1 — update all counters based on the result
        self.played += 1
        self.goals_for += scored
        self.goals_against += conceded
        if scored > conceded:
            self.wins += 1
        elif scored == conceded:
            self.draws += 1
        else:
            self.losses += 1


# Quick test — one win (2-1), one draw (0-0), one loss (0-1).
# Once Task 1 is done, this should print:  1 1 1   then   Points: 4  GD: 0
t = Team("Test")

t.add_match(2, 1)
````
