## Task 2 — Add derived columns

Add these new columns to `df`:

- `points_plus_assists` = points + assists
- `minutes_per_game` = minutes / games_played
- `pts_per_36` = points / (minutes / 36)
- `pa_per_36` = points_plus_assists / (minutes / 36)
- `is_starter` = `True` if minutes_per_game ≥ 24, else `False`

Show the head of the DataFrame with these new columns.

````python
# TODO Task 2
df["points_plus_assists"] = df["points"] + df["assists"]
df["minutes_per_game"] = df["minutes"] / df["games_played"]
df["pts_per_36"] = df["points"] / (df["minutes"]/36)
df["pa_per_36"] = df["points_plus_assists"] / (df["minutes"]/36)
df["is_starter"] = df["minutes_per_game"] >= 24
df.loc[1]
````

## Task 3 — Build a shortlist of scoring targets

Create a DataFrame `scoring_targets` keeping only players who:

- Play a perimeter position (PG, SG or SF)
- Are starters (per Task 2)
- Have at least 22.0 combined points+assists per 36 minutes

Sort by `pa_per_36` descending. Show only the columns: `name`, `position`, `team`, `age`, `points`, `assists`, `minutes`, `pa_per_36`.

````python
# TODO Task 3
scoring = df[(df["position"].isin(["PG", "SG", "SF"])) & (df["is_starter"] == True) & (df["pa_per_36"] >= 22)]
scoring_targets = scoring.sort_values("pa_per_36", ascending=False)[["name", "position", "team", "age", "points", "assists", "minutes", "is_starter", "pa_per_36"]]
scoring_targets
````
