## Starting data

Do not modify this cell. These are the data points you will analyse.

````python
team = "Lagos FC"

# Goals scored by the team in each of 15 matches
goals_scored   = [2, 1, 3, 0, 4, 2, 1, 1, 0, 3, 2, 1, 0, 2, 5]

# Goals conceded by the team in each of 15 matches
goals_conceded = [1, 1, 2, 0, 2, 0, 1, 2, 3, 2, 2, 1, 1, 2, 1]
````

---

## Task 1 — Basic statistics

Calculate and print:

1. Number of matches played
2. Total goals scored
3. Total goals conceded
4. Goal difference (scored − conceded)

Hint: `sum()` and `len()` are your friends.

> ✅ **Self-check (Task 1).** With this team's data, a correct Task 1 prints:
> **Matches played: 15 · Goals scored: 27 · Goals conceded: 21 · Goal difference: +6**.
> If your numbers differ, fix Task 1 before continuing — every later task builds on these.

````python
# TODO Task 1

matches_played   = len(goals_scored)
total_scored     = sum ((goals_scored))
total_conceded   = sum ((goals_conceded))
goal_difference  = total_scored - total_conceded

print(f"Matches played:  {matches_played}")
print(f"Goals scored:    {total_scored}")
print(f"Goals conceded:  {total_conceded}")
print(f"Goal difference: {goal_difference:+d}")
````

## Task 2 — Per-match averages

1. Average goals scored per match (2 decimals)
2. Average goals conceded per match (2 decimals)

````python
# TODO Task 2

avg_scored   = total_scored / len(goals_scored)
avg_conceded = total_conceded / len(goals_conceded)

print(f"Average goals for:     {avg_scored:.2f}")
print(f"Average goals against: {avg_conceded:.2f}")
````

## Task 3 — Match results

Loop through the 15 matches. For each one, determine if it was a **win (W)**, **draw (D)** or **loss (L)**. Count the totals.

Hint: loop over `range(len(goals_scored))` so you can access both lists with the same index `i`.

````python
# TODO Task 3

wins = 0
draws = 0
losses = 0

# Hint:
for i in range(len(goals_scored)):
    scored = goals_scored[i]
    conceded = goals_conceded[i]
    if scored > conceded:
        wins += 1
    elif scored == conceded:
        draws += 1
    else:
        losses += 1

print(f"Wins:   {wins}")
print(f"Draws:  {draws}")
print(f"Losses: {losses}")
````

## Task 4 — Points and win percentage

In most leagues, points are distributed as:

- Win: 3 points
- Draw: 1 point
- Loss: 0 points

Calculate:

1. Total team points
2. Points per match (2 decimals)
3. Win percentage (2 decimals)

````python
# TODO Task 4

nb_win = 6
nb_draw = 6
nb_loss = 3

total_points     = (nb_win * 3) + (nb_draw * 1) + (nb_loss * 0)
points_per_match = total_points / len(goals_scored)
win_percentage   = nb_win *100 / len(goals_scored)   # as a percentage (e.g. 40.00), not a fraction (0.40)

print(f"Total points:     {total_points}")
print(f"Points per match: {points_per_match:.2f}")
print(f"Win percentage:   {win_percentage:.2f}%")
````
