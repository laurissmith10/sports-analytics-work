## Season analysis of 5 sprinters

You are given a 2D array of 100m times. Each row is a sprinter; each column is one of 8 races across the season. Some entries are `np.nan` — the sprinter did not race that day.

Your job: produce a season report using vectorised NumPy operations (no Python loops over individual values).

**Estimated time:** ~1 hour.

---

````python
import numpy as np

# Athletes (rows) × 8 races (columns).  np.nan = did not race.
athlete_names = np.array(["Okocha", "Mensah", "Ibrahim", "Asante", "Boateng"])

times = np.array([
    [10.45, 10.31, 10.28, 10.22, 10.18, np.nan, 10.15, 10.19],   # Okocha
    [11.02, 10.85, np.nan, 10.71, 10.65, 10.69, 10.55, np.nan],  # Mensah
    [10.89, 10.71, 10.65, 10.50, np.nan, 10.42, 10.38, 10.31],   # Ibrahim
    [10.62, 10.55, 10.51, np.nan, 10.45, 10.40, 10.36, 10.30],   # Asante
    [np.nan, 10.92, 10.85, 10.80, 10.75, 10.71, 10.65, 10.60],   # Boateng
])

print("shape:", times.shape)
print(times)
````

````python
mask = (~np.isnan(times)).sum(axis=1)
print(mask)
````

## Task 1 — How many races did each athlete actually run?

Return a 1D array `races_run` with one value per athlete: the count of non-nan entries in their row.

Hint: `(~np.isnan(row)).sum()` for a single row — the parentheses around `~np.isnan(row)` matter, because `~` must flip the mask *before* you count. To do all rows at once: use `.sum(axis=...)`.

````python
# TODO Task 1
races_run = (~np.isnan(times)).sum(axis=1)

if races_run is None:
    print("Task 1 — not filled in yet.")
else:
    for name, count in zip(athlete_names, races_run):
        print(f"{name}: {count} races")
````

## Task 2 — Best and worst time per athlete

Compute:
- `best_times` — best (lowest) time per athlete, ignoring nans
- `worst_times` — worst (highest) time per athlete, ignoring nans

Hint: `np.nanmin` and `np.nanmax` with the right axis.

````python
# TODO Task 2
best_times = np.nanmin(times, axis=1)
worst_times = np.nanmax(times, axis=1)

if best_times is None or worst_times is None:
    print("Task 2 — not filled in yet.")
else:
    for i, name in enumerate(athlete_names):
        print(f"{name}: best {best_times[i]:.2f} / worst {worst_times[i]:.2f}")
````

## Task 3 — Season average and consistency

For each athlete, compute their season mean (`means`) and standard deviation (`stds`), ignoring nans.

The most **consistent** athlete is the one with the lowest standard deviation. Find them and print their name.

````python
# TODO Task 3
means = np.nanmean(times, axis=1)
stds = np.nanstd(times, axis=1)
most_consistent_index = np.nanargmin(stds)

if most_consistent_index is None:
    print("Task 3 — not filled in yet.")
else:
    print(f"Most consistent athlete: {athlete_names[most_consistent_index]}")
````

## Task 4 — Top marks

Count how many runs each athlete had **below 10.30s** (considered top marks).

Hint: a 2D boolean mask + sum along the right axis. Nans in the comparison are treated as `False` automatically.

````python
# TODO Task 4
top_marks_per_athlete = (times < 10.30).sum(axis=1)

if top_marks_per_athlete is None:
    print("Task 4 — not filled in yet.")
else:
    for i, name in enumerate(athlete_names):
        print(f"{name}: {top_marks_per_athlete[i]} top marks (< 10.30s)")
````

## Task 5 — Race-by-race field analysis

For each race (each column), compute:
- The average time of all athletes who ran it
- The fastest time of the race

Hint: aggregations along `axis=0` collapse rows → one value per column.

````python
# TODO Task 5
race_means = np.nanmean(times, axis=0)
race_winners = np.nanmin(times, axis=0)

if race_means is None or race_winners is None:
    print("Task 5 — not filled in yet.")
else:
    for race_idx in range(times.shape[1]):
        print(f"Race {race_idx + 1}: avg {race_means[race_idx]:.2f}, "
              f"best {race_winners[race_idx]:.2f}")
````

## Bonus (optional) — Improvement across the season

For each athlete, compute **last valid time minus first valid time**. A negative number means they improved (they got faster across the season).

It is trickier than it looks because of nans. Hint: filter the row to non-nan values, then take [0] and [-1].

This bonus probably requires a Python loop — that is fine, NumPy aggregations don't directly give you "first non-nan" across rows.

````python
# Bonus

for i, name in enumerate(athlete_names):
    valid = times[i][~np.isnan(times[i])]
    # your code here
    improvement = valid[-1] - valid[0]
    print(f"Athlete {name}: improvement {improvement:+.2f}s")
````
