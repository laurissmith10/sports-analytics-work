## Task 6 — Edit, commit and push your first change

This is the loop you will repeat every week of the course. Practise it now.

**Before you start:** make sure you already ran `gh auth login` (Part 2 of the theory guide). Do it now if you have not — otherwise Step 5 below will fail or hang asking for a password you don't have.

**Step 1 — Edit the README**

Open the `README.md` file in your `sports-analytics-work` folder using VSCode. Replace its content with something like:

```
# Sports Analytics Work

This repository holds my exercises, notes and projects for the ASE Athletics Sports Data Analytics & BI program (2026/27 cohort).

Week 01: Setup complete.
```

Save the file (`Ctrl + S` or `Cmd + S`).

**Step 2 — Check what changed**

In your terminal (still inside `sports-analytics-work`):

```
git status
```

You should see `README.md` listed as modified.

**Step 3 — Stage the change**

```
git add README.md
```

**Step 4 — Commit the change**

```
git commit -m "Update README with course description"
```

**Step 5 — Push to GitHub**

```
git push
```

**Step 6 — Verify on GitHub**

Reload your repository page on GitHub. You should see your updated README at the top.

## Bonus (optional) — VSCode Git integration
