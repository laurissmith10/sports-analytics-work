```python
goals = [2, 1, 3, 0, 4, 2, 1]

print(goals[0:3])   # [2, 1, 3]   first three (0, 1, 2 - not 3)
print(goals[:3])    # [2, 1, 3]   same thing, start is optional
print(goals[-3:])   # [4, 2, 1]   the last three - recent form
```

```python
goals = [2, 1, 3, 0, 4, 2, 1]
#         0   1   2   3   4   5   6      <- the positions

print(goals[::1])    # [2, 1, 3, 0, 4, 2, 1]   step 1 = every match (the default)
print(goals[::2])    # [2, 3, 4, 1]            every OTHER match: positions 0, 2, 4, 6
print(goals[1::2])   # [1, 0, 2]               start at 1, then every other: 1, 3, 5
print(goals[::-1])   # [1, 2, 4, 0, 3, 1, 2]   a NEGATIVE step walks backwards
```
